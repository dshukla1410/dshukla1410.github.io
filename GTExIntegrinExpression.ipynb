```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from google.colab import files

sns.set_theme(style="whitegrid")

uploaded = files.upload()

file_name = list(uploaded.keys())[0]

df = pd.read_excel(file_name)

print("Dataset shape:", df.shape)

display(df.head())
print("Column names:")
print(df.columns.tolist())

print("\nData types:")
print(df.dtypes)

print("\nTissue categories:")
print(df["primary_site"].unique())

# Rename columns for easier use
df = df.rename(columns={
    "Unnamed: 0": "sample_id",
    "primary_site": "Tissue"
})

# Identify gene columns
id_cols = ["sample_id", "Tissue"]

gene_cols = [col for col in df.columns if col not in id_cols]

print("Number of integrin genes:", len(gene_cols))

print("Integrin genes:")
print(gene_cols)

# Select only Lung and Breast samples
tissue_data = df[df["Tissue"].isin(["Lung", "Breast"])].copy()

print("Number of samples in each tissue:")
print(tissue_data["Tissue"].value_counts())

display(tissue_data.head())

long_df = tissue_data.melt(
    id_vars=["sample_id", "Tissue"],
    value_vars=gene_cols,
    var_name="Gene",
    value_name="Expression"
)

print("Reshaped dataset:")
display(long_df.head())

print("New dataset shape:", long_df.shape)

summary = (
    long_df.groupby(["Gene", "Tissue"])["Expression"]
    .agg(["count", "mean", "median", "std"])
    .reset_index()
)

display(summary)

plt.figure(figsize=(8, 5))

sns.violinplot(
    data=long_df,
    x="Tissue",
    y="Expression",
    inner="quartile",
    cut=0
)

plt.title("Integrin Expression Distributions: Lung vs Breast")
plt.xlabel("Tissue")
plt.ylabel("Expression")

plt.tight_layout()
plt.show()

selected_gene = "ITGAX"

gene_data = long_df[long_df["Gene"] == selected_gene].copy()

print("Selected gene:", selected_gene)

gene_summary = gene_data.groupby("Tissue")["Expression"].agg(
    ["count", "mean", "median", "std"]
)

display(gene_summary)

plt.figure(figsize=(7, 5))

sns.boxplot(
    data=gene_data,
    x="Tissue",
    y="Expression",
    showfliers=False
)

sns.stripplot(
    data=gene_data,
    x="Tissue",
    y="Expression",
    color="black",
    alpha=0.2,
    size=2
)

plt.title("ITGAX Expression: Lung vs Breast")
plt.xlabel("Tissue")
plt.ylabel("Expression")

plt.tight_layout()
plt.show()

# Save overall distribution plot
plt.figure(figsize=(8, 5))

sns.violinplot(
    data=long_df,
    x="Tissue",
    y="Expression",
    inner="quartile",
    cut=0
)

plt.title("Integrin Expression Distributions: Lung vs Breast")
plt.xlabel("Tissue")
plt.ylabel("Expression")

plt.tight_layout()

plt.savefig(
    "integrin_lung_breast_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()


# Save ITGAX plot
plt.figure(figsize=(7, 5))

sns.boxplot(
    data=gene_data,
    x="Tissue",
    y="Expression",
    showfliers=False
)

sns.stripplot(
    data=gene_data,
    x="Tissue",
    y="Expression",
    color="black",
    alpha=0.2,
    size=2
)

plt.title("ITGAX Expression: Lung vs Breast")
plt.xlabel("Tissue")
plt.ylabel("Expression")

plt.tight_layout()

plt.savefig(
    "ITGAX_lung_breast.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

The overall distribution of integrin gene expression was higher in Lung tissue compared to Breast tissue, with median expression values of approximately 4.45 and 2.93, respectively. The selected gene, ITGAX, also showed higher median expression in Lung (6.82) than in Breast (2.43). These differences may suggest tissue-specific variation in cell composition and cell–matrix interactions, as integrins are involved in cell adhesion and communication with the extracellular matrix. However, these results do not establish the exact biological mechanism responsible for the differences. One limitation is that the dataset contains bulk tissue measurements, which combine different cell types and may be influenced by variation between individual donors.
