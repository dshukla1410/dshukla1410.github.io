```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x=np.linspace (0,10,100)
y=x**3
fig, ax = plt.subplots()
ax.plot (x,y)
ax.set_xlabel("x")
ax.set_ylabel("y")
plt.show()
```
