```python

# ============================================================
# Logistic Regression with Integrin Expression
# Binary Classification: Lung vs Brain
# Multi-Class Classification: All 7 Organs
# ============================================================

# 1. IMPORT LIBRARIES
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler, label_binarize
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    confusion_matrix,
    ConfusionMatrixDisplay,
    classification_report,
    roc_curve,
    roc_auc_score
)

# ============================================================
# 2. LOAD DATASET
# ============================================================

url = "https://raw.githubusercontent.com/yongmeiwang/AI_BigData_CancerBiology/main/Data/gtex_integrin_7_organs.xlsx"

Gtex_integrin = pd.read_excel(url)

print("Dataset loaded successfully!")
print("Dataset shape:", Gtex_integrin.shape)
display(Gtex_integrin.head())

print("\nColumns:")
print(Gtex_integrin.columns.tolist())

print("\nOrgan counts:")
print(Gtex_integrin["primary_site"].value_counts())

# ============================================================
# 3. IDENTIFY INTEGRIN GENE COLUMNS
# ============================================================

# Select numeric columns, excluding numeric metadata if present.
# Integrin expression columns are expected to be numeric.
gene_cols = Gtex_integrin.select_dtypes(
    include=np.number
).columns.tolist()

# Remove columns that are constant or entirely missing.
gene_cols = [
    col for col in gene_cols
    if Gtex_integrin[col].notna().any()
    and Gtex_integrin[col].nunique(dropna=True) > 1
]

if len(gene_cols) < 2:
    raise ValueError(
        "At least two numeric gene-expression columns are required."
    )

print("\nNumeric candidate gene columns:")
print(gene_cols)

# IMPORTANT:
# Check the printed columns. If any are sample IDs or metadata,
# remove them from gene_cols and keep only integrin expression genes.

# ============================================================
# PART 1: VISUAL EXPLORATION
# ============================================================

# Compare Lung vs Brain using the supplied primary_site column.
Gtex_lung_brain = Gtex_integrin[
    Gtex_integrin["primary_site"].isin(["Lung", "Brain"])
].copy()

print("\nLung vs Brain sample counts:")
print(Gtex_lung_brain["primary_site"].value_counts())

if Gtex_lung_brain["primary_site"].nunique() != 2:
    raise ValueError("Both Lung and Brain samples must be present.")

# Plot up to 8 candidate integrin genes.
plot_genes = gene_cols[:8]

plot_df = Gtex_lung_brain[
    ["primary_site"] + plot_genes
].melt(
    id_vars="primary_site",
    var_name="Integrin_Gene",
    value_name="Expression"
).dropna()

plt.figure(figsize=(max(12, len(plot_genes) * 2), 6))

sns.violinplot(
    data=plot_df,
    x="Integrin_Gene",
    y="Expression",
    hue="primary_site",
    split=True,
    inner="quartile",
    cut=0
)

plt.title("Integrin Expression: Lung vs Brain")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# ============================================================
# 4. IDENTIFY STRONG AND WEAK PREDICTORS
# ============================================================

# Brain = 0, Lung = 1
y_all_binary = (
    Gtex_lung_brain["primary_site"] == "Lung"
).astype(int)

auc_results = []

for gene in gene_cols:
    temp = pd.DataFrame({
        "expression": Gtex_lung_brain[gene],
        "target": y_all_binary
    }).dropna()

    if temp["target"].nunique() < 2:
        continue

    raw_auc = roc_auc_score(
        temp["target"],
        temp["expression"]
    )

    auc_results.append({
        "Gene": gene,
        "Raw_AUC": raw_auc,
        "Separation_AUC": max(raw_auc, 1 - raw_auc)
    })

auc_table = pd.DataFrame(auc_results).sort_values(
    "Separation_AUC",
    ascending=False
).reset_index(drop=True)

print("\nGene separation ranking:")
display(auc_table)

if len(auc_table) < 2:
    raise ValueError("Not enough genes to compare.")

strong_gene = auc_table.iloc[0]["Gene"]
weak_gene = auc_table.iloc[-1]["Gene"]

print("\nStrong predictor:", strong_gene)
print("Exploratory separation AUC:",
      round(auc_table.iloc[0]["Separation_AUC"], 3))

print("\nWeak predictor:", weak_gene)
print("Exploratory separation AUC:",
      round(auc_table.iloc[-1]["Separation_AUC"], 3))

# ============================================================
# PART 2 AND 3: BINARY LOGISTIC REGRESSION
# ============================================================

def run_binary_model(gene, random_state=42):
    data = Gtex_lung_brain[
        [gene, "primary_site"]
    ].dropna().copy()

    X = data[[gene]]
    y = (data["primary_site"] == "Lung").astype(int)

    X_train, X_test, y_train, y_test = train_test_split(
        X,
        y,
        test_size=0.30,
        random_state=random_state,
        stratify=y
    )

    model = make_pipeline(
        StandardScaler(),
        LogisticRegression(max_iter=2000)
    )

    model.fit(X_train, y_train)

    predictions = model.predict(X_test)
    probabilities = model.predict_proba(X_test)[:, 1]

    accuracy = accuracy_score(y_test, predictions)
    cm = confusion_matrix(y_test, predictions, labels=[0, 1])
    auc = roc_auc_score(y_test, probabilities)

    print("\n" + "=" * 55)
    print("GENE:", gene)
    print("Random state:", random_state)
    print(f"Accuracy: {accuracy:.3f}")
    print(f"ROC-AUC: {auc:.3f}")
    print("Confusion matrix (Brain, Lung):")
    print(cm)

    print("\nClassification report:")
    print(classification_report(
        y_test,
        predictions,
        labels=[0, 1],
        target_names=["Brain", "Lung"],
        zero_division=0
    ))

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    ConfusionMatrixDisplay(
        confusion_matrix=cm,
        display_labels=["Brain", "Lung"]
    ).plot(ax=axes[0], colorbar=False)

    axes[0].set_title(f"Confusion Matrix: {gene}")

    fpr, tpr, _ = roc_curve(y_test, probabilities)

    axes[1].plot(
        fpr, tpr,
        label=f"AUC = {auc:.3f}"
    )
    axes[1].plot([0, 1], [0, 1], "k--")
    axes[1].set_xlabel("False Positive Rate")
    axes[1].set_ylabel("True Positive Rate")
    axes[1].set_title(f"ROC Curve: {gene}")
    axes[1].legend()

    plt.tight_layout()
    plt.show()

    return {
        "gene": gene,
        "random_state": random_state,
        "accuracy": accuracy,
        "auc": auc,
        "confusion_matrix": cm
    }

# Strong predictor
strong_results = run_binary_model(
    strong_gene,
    random_state=42
)

# Weak predictor
weak_results = run_binary_model(
    weak_gene,
    random_state=42
)

# Compare the strong and weak models
binary_comparison = pd.DataFrame([
    {
        "Model": "Strong predictor",
        "Gene": strong_gene,
        "Accuracy": strong_results["accuracy"],
        "AUC": strong_results["auc"]
    },
    {
        "Model": "Weak predictor",
        "Gene": weak_gene,
        "Accuracy": weak_results["accuracy"],
        "AUC": weak_results["auc"]
    }
])

print("\nStrong vs Weak Model Comparison:")
display(binary_comparison.round(3))

# ============================================================
# PART 4: MODEL STABILITY
# ============================================================

# Repeat the strong model with a different random state.
stability_results = run_binary_model(
    strong_gene,
    random_state=7
)

stability_comparison = pd.DataFrame([
    {
        "Random State": 42,
        "Accuracy": strong_results["accuracy"],
        "AUC": strong_results["auc"]
    },
    {
        "Random State": 7,
        "Accuracy": stability_results["accuracy"],
        "AUC": stability_results["auc"]
    }
])

print("\nModel Stability Comparison:")
display(stability_comparison.round(3))

print("Confusion matrix, random_state=42:")
print(strong_results["confusion_matrix"])

print("\nConfusion matrix, random_state=7:")
print(stability_results["confusion_matrix"])

# ============================================================
# PART 5: MULTI-CLASS LOGISTIC REGRESSION
# ALL SEVEN ORGANS, USING TWO INTEGRIN GENES
# ============================================================

multi_df = Gtex_integrin.dropna(
    subset=["primary_site"]
).copy()

# Select two variable numeric features as a starting point.
# Verify these are integrin expression genes, not metadata.
multi_variances = multi_df[gene_cols].var().sort_values(
    ascending=False
)

selected_genes = multi_variances.index[:2].tolist()

print("\nSelected multiclass genes:", selected_genes)
print("\nOrgan counts:")
print(multi_df["primary_site"].value_counts())

multi_data = multi_df[
    ["primary_site"] + selected_genes
].dropna().copy()

X_multi = multi_data[selected_genes]
y_multi = multi_data["primary_site"].astype(str)

print("\nNumber of organ classes:", y_multi.nunique())
print("Classes:", sorted(y_multi.unique()))

X_train_multi, X_test_multi, y_train_multi, y_test_multi = (
    train_test_split(
        X_multi,
        y_multi,
        test_size=0.30,
        random_state=42,
        stratify=y_multi
    )
)

multi_model = make_pipeline(
    StandardScaler(),
    LogisticRegression(max_iter=3000)
)

multi_model.fit(X_train_multi, y_train_multi)

multi_predictions = multi_model.predict(X_test_multi)

multi_accuracy = accuracy_score(
    y_test_multi,
    multi_predictions
)

organ_labels = multi_model.classes_

multi_cm = confusion_matrix(
    y_test_multi,
    multi_predictions,
    labels=organ_labels
)

print("\nMulti-class accuracy:", round(multi_accuracy, 3))
print("\nMulti-class confusion matrix:")
print("Row/column order:", list(organ_labels))
print(multi_cm)

print("\nPer-organ classification report:")
print(classification_report(
    y_test_multi,
    multi_predictions,
    labels=organ_labels,
    zero_division=0
))

# Plot the multi-class confusion matrix
plt.figure(figsize=(10, 8))

sns.heatmap(
    multi_cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=organ_labels,
    yticklabels=organ_labels
)

plt.xlabel("Predicted organ")
plt.ylabel("Actual organ")
plt.title("Multi-Class Logistic Regression: All Organs")
plt.xticks(rotation=45, ha="right")
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

# ============================================================
# OPTIONAL: MULTI-CLASS ROC-AUC
# ============================================================

if len(organ_labels) >= 3:
    y_test_binarized = label_binarize(
        y_test_multi,
        classes=organ_labels
    )

    multi_probabilities = multi_model.predict_proba(
        X_test_multi
    )

    multi_auc = roc_auc_score(
        y_test_binarized,
        multi_probabilities,
        multi_class="ovr",
        average="macro"
    )

    print("\nMulti-class macro ROC-AUC:", round(multi_auc, 3))
else:
    multi_auc = np.nan

# ============================================================
# FINAL RESULTS SUMMARY
# ============================================================

final_summary = pd.DataFrame([
    {
        "Analysis": "Binary - Strong",
        "Gene(s)": strong_gene,
        "Accuracy": strong_results["accuracy"],
        "AUC": strong_results["auc"]
    },
    {
        "Analysis": "Binary - Weak",
        "Gene(s)": weak_gene,
        "Accuracy": weak_results["accuracy"],
        "AUC": weak_results["auc"]
    },
    {
        "Analysis": "Stability - New Split",
        "Gene(s)": strong_gene,
        "Accuracy": stability_results["accuracy"],
        "AUC": stability_results["auc"]
    },
    {
        "Analysis": "Multiclass - All Organs",
        "Gene(s)": ", ".join(selected_genes),
        "Accuracy": multi_accuracy,
        "AUC": multi_auc
    }
])

print("\nFINAL RESULTS SUMMARY:")
display(final_summary.round(3))

# Save results as a CSV file
final_summary.to_csv(
    "integrin_logistic_regression_results.csv",
    index=False
)

print("\nResults saved to integrin_logistic_regression_results.csv")
```
